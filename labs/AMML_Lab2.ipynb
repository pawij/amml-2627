Supervised contrastive learning
===

Now we're going to look at using cross-entropy in a deep learning setting of supervised contrastive learning.

Have a look at this paper for the background theory to our particular implementation of supervised contrastive learning.

https://arxiv.org/pdf/2004.11362

The following code implements these steps:

* Computes CIFAR-10 normalization statistics automatically (no magic numbers).
* Uses MobileNetV2 as the encoder.
* Uses a SupCon loss that excludes self-pairs.
* Pretrains the encoder using supervised contrastive learning.
* Freezes the encoder.
* Trains a classifier head using cross-entropy loss.
* Evaluates test accuracy.
* Visualizes embeddings with t-SNE.

Import the libraries that we will need for the analysis.

```python
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F

from torchvision import datasets
from torchvision import transforms
from torchvision.models import mobilenet_v2

from torch.utils.data import DataLoader

from sklearn.manifold import TSNE
from sklearn.metrics import accuracy_score

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)
```

Fixed constants (aka "hyperparameters").

```python
BATCH_SIZE = 256

LEARNING_RATE = 1e-3

PROJECTION_DIM = 128

HIDDEN_DIM = 512

NUM_CLASSES = 10

TEMPERATURE = 0.07

MAX_ROTATION_DEGREES = 7.2

CONTRASTIVE_EPOCHS = 20

CLASSIFIER_EPOCHS = 20
```

Download training data.

```python
raw_train = datasets.CIFAR10(
    root="./data",
    train=True,
    download=True
)

raw_test = datasets.CIFAR10(
    root="./data",
    train=False,
    download=True
)

print("Training examples:", len(raw_train))
print("Test examples:", len(raw_test))
```

Compute dataset statistics.

```python
x = raw_train.data.astype(np.float32) / 255.0

mean = x.mean(axis=(0,1,2))
std = x.std(axis=(0,1,2))

print("Mean =", mean)
print("Std  =", std)
```

Data augmentation. Note this is an essential step for contrastive learning!

```python
train_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(
        MAX_ROTATION_DEGREES
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=mean.tolist(),
        std=std.tolist()
    )
])

test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=mean.tolist(),
        std=std.tolist()
    )
])
```

Create final datasets.

```python
train_dataset = datasets.CIFAR10(
    root="./data",
    train=True,
    download=False,
    transform=train_transform
)

test_dataset = datasets.CIFAR10(
    root="./data",
    train=False,
    download=False,
    transform=test_transform
)
```

Create data loaders.

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2
)
```

Visualise data - good practice for any ML engineer!

```python
images, labels = next(iter(train_loader))

fig, axes = plt.subplots(2,5, figsize=(10,4))

for i in range(10):

    r = i // 5
    c = i % 5

    img = images[i]

    img = img.permute(1,2,0).numpy()

    img = img * std + mean

    img = np.clip(img,0,1)

    axes[r,c].imshow(img)
    axes[r,c].set_title(str(labels[i].item()))
    axes[r,c].axis("off")

plt.show()
```

Define encoder class.

```python
class Encoder(nn.Module):

    def __init__(self):

        super().__init__()

        backbone = mobilenet_v2(
            weights=None
        )

        backbone.classifier = nn.Identity()

        self.backbone = backbone

        self.output_dim = 1280

    def forward(self,x):

        return self.backbone(x)
```

Projection head. In SimCLR contrastive learning, the representation and loss are in different spaces.

```python
class ProjectionHead(nn.Module):

    def __init__(
        self,
        encoder,
        projection_dim=128
    ):

        super().__init__()

        self.encoder = encoder

        self.projector = nn.Sequential(

            nn.Linear(
                encoder.output_dim,
                projection_dim
            ),

            nn.ReLU()
        )

    def forward(self,x):

        h = self.encoder(x)

        z = self.projector(h)

        return z
```

Supervised contrastive loss, corresponding to the "SupCon" model introduced in the paper above. Note self-pairs are excluded, why?


```python
class SupConLoss(nn.Module):

    def __init__(
        self,
        temperature=0.07
    ):
        super().__init__()

        self.temperature = temperature

    def forward(
        self,
        features,
        labels
    ):

        features = F.normalize(
            features,
            dim=1
        )

        similarities = (
            torch.matmul(
                features,
                features.T
            )
            / self.temperature
        )

        batch_size = labels.shape[0]

        labels = labels.view(-1,1)

        positive_mask = (
            labels == labels.T
        )

        diagonal_mask = torch.eye(
            batch_size,
            dtype=torch.bool,
            device=features.device
        )

        positive_mask = (
            positive_mask
            &
            (~diagonal_mask)
        )

        similarities = similarities.masked_fill(
            diagonal_mask,
            -1e9
        )

        log_prob = (
            similarities
            -
            torch.logsumexp(
                similarities,
                dim=1,
                keepdim=True
            )
        )

        positive_mask = positive_mask.float()

        positive_count = positive_mask.sum(1)

        valid_rows = positive_count > 0

        mean_log_prob = (
            positive_mask * log_prob
        ).sum(1)

        mean_log_prob = (
            mean_log_prob
            /
            (positive_count + 1e-8)
        )

        loss = -mean_log_prob[valid_rows]

        return loss.mean()
```

Create contrastive model.

```python
encoder = Encoder()

contrastive_model = ProjectionHead(
    encoder,
    PROJECTION_DIM
).to(device)

contrastive_loss = SupConLoss(
    TEMPERATURE
)

optimizer = torch.optim.Adam(
    contrastive_model.parameters(),
    lr=LEARNING_RATE
)
```

Contrastive pretraining.

```python
for epoch in range(CONTRASTIVE_EPOCHS):

    contrastive_model.train()

    running_loss = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        z = contrastive_model(images)

        loss = contrastive_loss(
            z,
            labels
        )

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

    print(
        f"Epoch {epoch+1:02d} "
        f"Loss={running_loss/len(train_loader):.4f}"
    )
```

Classifier.

```python
class Classifier(nn.Module):

    def __init__(
        self,
        encoder,
        hidden_dim=512,
        num_classes=10
    ):

        super().__init__()

        self.encoder = encoder

        self.classifier = nn.Sequential(

            nn.Dropout(0.5),

            nn.Linear(
                encoder.output_dim,
                hidden_dim
            ),

            nn.ReLU(),

            nn.Dropout(0.5),

            nn.Linear(
                hidden_dim,
                num_classes
            )
        )

    def forward(self,x):

        h = self.encoder(x)

        logits = self.classifier(h)

        return logits
```

Freeze encoder.

```python
for parameter in encoder.parameters():

    parameter.requires_grad = False
```

Create classifier - here's the cross entropy loss!

```python
classifier = Classifier(
    encoder,
    hidden_dim=HIDDEN_DIM,
    num_classes=NUM_CLASSES
).to(device)

criterion = nn.CrossEntropyLoss()

optimizer_cls = torch.optim.Adam(
    filter(
        lambda p: p.requires_grad,
        classifier.parameters()
    ),
    lr=LEARNING_RATE
)
```

Train classifier.

```python
for epoch in range(CLASSIFIER_EPOCHS):

    classifier.train()

    running_loss = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer_cls.zero_grad()

        logits = classifier(images)

        loss = criterion(
            logits,
            labels
        )

        loss.backward()

        optimizer_cls.step()

        running_loss += loss.item()

    print(
        f"Epoch {epoch+1:02d} "
        f"Loss={running_loss/len(train_loader):.4f}"
    )
```

Evaluation.

```python
classifier.eval()

preds = []
targets = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)

        logits = classifier(images)

        pred = logits.argmax(1)

        preds.extend(pred.cpu().numpy())
        targets.extend(labels.numpy())

acc = accuracy_score(
    targets,
    preds
)

print(
    f"Test accuracy = "
    f"{100*acc:.2f}%"
)
```

Extract embeddings.

```python
encoder.eval()

features = []
labels_all = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)

        h = encoder(images)

        features.append(
            h.cpu().numpy()
        )

        labels_all.append(
            labels.numpy()
        )

features = np.concatenate(features)

labels_all = np.concatenate(labels_all)

print(features.shape)
```

Visualise embeddings in latent space.

```python
idx = np.random.choice(
    len(features),
    2000,
    replace=False
)

X = features[idx]

y = labels_all[idx]

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_2d = tsne.fit_transform(X)

plt.figure(figsize=(8,8))

scatter = plt.scatter(
    X_2d[:,0],
    X_2d[:,1],
    c=y,
    s=6,
    cmap="tab10"
)

plt.colorbar(scatter)

plt.title(
    "Supervised Contrastive Representations"
)

plt.show()
```
